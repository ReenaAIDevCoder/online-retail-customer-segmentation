# Step 4 — Duplicate Records

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

# Step 5 — InvoiceDate Convert

```python
df = df.copy()

df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)

print(df["InvoiceDate"].dtype)
```

```python
df = df.drop_duplicates()
```

```python
df = df[df["Quantity"] > 0]
```

```python
cancelled_orders = df[
    df["InvoiceNo"].astype(str).str.startswith("C")
]

print("Cancelled orders:", len(cancelled_orders))
```

# 1. Cancelled Orders Remove

```python
cancelled_orders = df[
    df["InvoiceNo"].astype(str).str.startswith("C")
]

print("Cancelled orders:", len(cancelled_orders))
```

```python
df = df[
    ~df["InvoiceNo"].astype(str).str.startswith("C")
].copy()

print("Shape after removing cancelled orders:", df.shape)
```

# 2. Check Invalid Quantity

```python
print("Quantity <= 0:", (df["Quantity"] <= 0).sum())
```

# 3. Remove Invalid Quantity

```python
df = df[df["Quantity"] > 0].copy()

print("Shape after removing invalid quantity:", df.shape)
```

# 4. Check Invalid Unit Price

```python
print("UnitPrice <= 0:", (df["UnitPrice"] <= 0).sum())
```

# 5. Remove Invalid Unit Price

```python
df = df[df["UnitPrice"] > 0].copy()

print("Shape after removing invalid prices:", df.shape)
```

# 6. Check Missing CustomerID

```python
print("Missing CustomerID:", df["CustomerID"].isnull().sum())
```

# 7. Remove Missing CustomerID

```python
df = df.dropna(subset=["CustomerID"]).copy()

print("Shape after removing missing CustomerID:", df.shape)
```

# 8. Convert CustomerID to Integer

```python
df["CustomerID"] = df["CustomerID"].astype(int)

print(df["CustomerID"].dtype)
```

# 9. Create Revenue

```python
df["Revenue"] = df["Quantity"] * df["UnitPrice"]

df[
    ["Quantity", "UnitPrice", "Revenue"]
].head()
```

# 10. Check Revenue

```python
print("Total Revenue:", round(df["Revenue"].sum(), 2))
print("Minimum Revenue:", df["Revenue"].min())
print("Maximum Revenue:", df["Revenue"].max())
```

# 11. Extract Year

```python
df["Year"] = df["InvoiceDate"].dt.year
```

# 12. Extract Month

```python
df["Month"] = df["InvoiceDate"].dt.month
```

# 13. Extract Day

```python
df["Day"] = df["InvoiceDate"].dt.day
```

# 14. Extract Hour

```python
df["Hour"] = df["InvoiceDate"].dt.hour
```

# 15. Extract Day of Week

```python
df["DayOfWeek"] = df["InvoiceDate"].dt.day_name()
```

# 16. Check New Features

```python
df[
    [
        "InvoiceDate",
        "Year",
        "Month",
        "Day",
        "Hour",
        "DayOfWeek",
        "Revenue"
    ]
].head()
```

# 17. Create Year-Month Feature

```python
df["YearMonth"] = df["InvoiceDate"].dt.to_period("M")

df[
    ["InvoiceDate", "YearMonth"]
].head()
```

```python
print("Final Dataset Shape:", df.shape)
```

# 24. Important Business Metrics

```python
total_revenue = df["Revenue"].sum()
total_customers = df["CustomerID"].nunique()
total_orders = df["InvoiceNo"].nunique()
total_products = df["StockCode"].nunique()
total_countries = df["Country"].nunique()

print("Total Revenue:", round(total_revenue, 2))
print("Total Customers:", total_customers)
print("Total Orders:", total_orders)
print("Total Products:", total_products)
print("Total Countries:", total_countries)
```

# 25. Average Order Value

```python
average_order_value = (
    df.groupby("InvoiceNo")["Revenue"]
    .sum()
    .mean()
)

print(
    "Average Order Value:",
    round(average_order_value, 2)
)
```

# 26. Check Date Range

```python
print("First Transaction:", df["InvoiceDate"].min())
print("Last Transaction:", df["InvoiceDate"].max())
```

# 27. Top 10 Countries by Revenue

```python
top_countries = (
    df.groupby("Country")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_countries
```

# 28. Top 10 Products by Revenue

```python
top_products = (
    df.groupby("Description")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products
```
